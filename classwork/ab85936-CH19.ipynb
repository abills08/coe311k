```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return -1000*y + 3000 - 2000 * np.exp(-1*x)

def euler(x_start, x_end, h):
    num_steps = int((x_end - x_start) / h)
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    y_values[0] = 0
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
    return x_values, y_values

h_values = [0.1,0.01,0.001]
x_start, x_end = 0, 30
for h in h_values:
    x1, y_approx1 = euler(x_start, x_end, h)
    plt.plot(x1, y_approx1, label=f'Euler\'s Method (h={h})')
x_exact = np.linspace(x_start, x_end, 1000)
y_exact = 3-0.998*np.exp(-1000*x_exact)-2.002*np.exp(-1*x_exact)
plt.plot(x_exact, y_exact, label = f'Exact Function y')
plt.xlabel('Time (t)')
plt.ylabel('Y')
plt.title('Euler\'s method vs Analytical Solution')
plt.legend()
plt.grid(True)
plt.show()

#step size wildly unstable with 0.1, 0.01
```

```python
#decided to run step size 0.001 by itself to see without the ones blowing up

x_start, x_end = 0, 30
x1, y_approx1 = euler(x_start, x_end, 0.001)
plt.plot(x1, y_approx1, label=f'Euler\'s Method (h=0.001)')
x_exact = np.linspace(x_start, x_end, 1000)
y_exact = 3-0.998*np.exp(-1000*x_exact)-2.002*np.exp(-1*x_exact)
plt.plot(x_exact, y_exact, label = f'Exact Function y')
plt.xlabel('Time (t)')
plt.ylabel('Y')
plt.title('Euler\'s method vs Analytical Solution')
plt.legend()
plt.grid(True)
plt.show()
```
