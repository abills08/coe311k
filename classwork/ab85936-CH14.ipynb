```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function and the domain
f = lambda x: np.sin(x)
a, b = 0, 2 * np.pi # Interval
N = 50 # Number of sample points

# Create a uniform grid
x = np.linspace(a, b, N)
y = f(x)

# Step size
h = (b - a) / (N - 1)

# Central Difference Approximation for the derivative
df_dxC = (f(x + h) - f(x - h)) / (2 * h)

# Forward Difference Approximation
df_dxF = (f(x+h)-f(x))/h

# Backward Difference Approximation
df_dxB = (f(x)-f(x-h))/h

# Plot the approximation and the true derivative
plt.plot(x, np.cos(x), color = 'black', label='Exact Derivative')
plt.plot(x, df_dxC, color = 'blue', label='Central Difference Approximation')
plt.plot(x, df_dxF, color = 'red', label='Forward Difference Approximation')
plt.plot(x, df_dxB, color = 'green', label='Backward Difference Approximation')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title("Approximating Derivative of f(x) = sin(x) using Central Difference")
plt.legend()
plt.show()
```
