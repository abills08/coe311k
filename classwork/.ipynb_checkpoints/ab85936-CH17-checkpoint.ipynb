```python
import numpy as np
import matplotlib.pyplot as plt

def eulers (h, k, x_start, x_end, y_0, v_0):
    # Create arrays to store x, y, and v values
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values)) 
    y_values[0] = y_0
    v_values[0] = v_0
    for i in range(1, len(x_values)):
        # Current values
        x_n = x_values[i - 1]
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]
        # Update equations
        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)
    return (x_values, y_values, v_values)

# Define the parameters
k = 1.0 # Constant in the equation (e.g., spring constant)
h = 0.01 # Step size
x_start = 0 # Start of the interval
x_end = 10 # End of the interval
y_0 = 1.0 # Initial condition for y
v_0 = 0.0 # Initial condition for dy/dx

# Plot the results
plt.figure(figsize=(10, 5))
x_values, y_values, v_values = eulers(h, k, x_start, x_end, y_0, v_0)
plt.plot(x_values, y_values, label='Position y(x) step size 0.01')
plt.plot(x_values, v_values, label='Velocity v(x) step size 0.01', linestyle='--')
h = 0.02
x_values, y_values, v_values = eulers(h, k, x_start, x_end, y_0, v_0)
plt.plot(x_values, y_values, label='Position y(x) step size 0.02')
plt.plot(x_values, v_values, label='Velocity v(x) step size 0.02', linestyle='--')
h = 0.03
x_values, y_values, v_values = eulers(h, k, x_start, x_end, y_0, v_0)
plt.plot(x_values, y_values, label='Position y(x) step size 0.03')
plt.plot(x_values, v_values, label='Velocity v(x) step size 0.03', linestyle='--')
h = 0.05
x_values, y_values, v_values = eulers(h, k, x_start, x_end, y_0, v_0)
plt.plot(x_values, y_values, label='Position y(x) step size 0.05')
plt.plot(x_values, v_values, label='Velocity v(x) step size 0.05', linestyle='--')
h = 0.1
x_values, y_values, v_values = eulers(h, k, x_start, x_end, y_0, v_0)
plt.plot(x_values, y_values, label='Position y(x) step size 0.1')
plt.plot(x_values, v_values, label='Velocity v(x) step size 0.1', linestyle='--')
h = 0.2
x_values, y_values, v_values = eulers(h, k, x_start, x_end, y_0, v_0)
plt.plot(x_values, y_values, label='Position y(x) step size 0.2')
plt.plot(x_values, v_values, label='Velocity v(x) step size 0.2', linestyle='--')
h = 0.3
x_values, y_values, v_values = eulers(h, k, x_start, x_end, y_0, v_0)
plt.plot(x_values, y_values, label='Position y(x) step size 0.3')
plt.plot(x_values, v_values, label='Velocity v(x) step size 0.3', linestyle='--')
h = 0.5
x_values, y_values, v_values = eulers(h, k, x_start, x_end, y_0, v_0)
plt.plot(x_values, y_values, label='Position y(x) step size 0.5')
plt.plot(x_values, v_values, label='Velocity v(x) step size 0.5', linestyle='--')
plt.xlabel('x')
plt.ylabel('y and v')
plt.title('Solution of d²y/dx² = -ky using Euler\'s Method')
plt.legend()
plt.grid(True)
plt.show()
```

The solution gets unstable quicker the larger the step size is. This begins to occur around x = 5, but it varries dependant on the step size being used. While the solutions typically still oscillate, the oscillations become much larger the larger the x value gets.

```python
k = 10.0
h = 0.01
x_values, y_values, v_values = eulers(h, k, x_start, x_end, y_0, v_0)
plt.plot(x_values, y_values, label='Position y(x) step size = 0.01, k = 10')
plt.plot(x_values, v_values, label='Velocity v(x) step size = 0.01, k = 10', linestyle='--')
plt.xlabel('x')
plt.ylabel('y and v')
plt.title('Solution of d²y/dx² = -ky using Euler\'s Method')
plt.legend()
plt.grid(True)
plt.show()
```

the velocity and position both continue to oscillate with larger and larger frequencies the longer we let it run. I do not think a larger h would solve this issue because as we saw earlier, that typically makes the system less stable and not more stable.
