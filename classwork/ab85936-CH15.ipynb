```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return -2 * y

def euler(x_start, x_end, h):
    num_steps = int((x_end - x_start) / h)
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    y_values[0] = 1
    error_y = np.zeros(len(y_values)-1)
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
        error_y[i] = y_values[i+1]-np.exp(-2 * x_values[i+1])
    plt.plot(x_values, y_values)
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(f"Approximation of dy/dx = -2y using Euler's Method with step size {h}")
    plt.show()
    total_error = sum(error_y)
    print(f"Error between points = {error_y}")
    print(f"Summed error = {total_error}")

step_sizes = np.arange(0.01, 1.01, 0.05)
x_start, x_end = 0, 2
for i in np.arange(0.01, 1.01, 0.05):
    euler(x_start, x_end, i)
```
