```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return -2 * y

def exactSolution(x_start, x_end, h, y0, coef):
    num_steps = int((x_end - x_start) / h) + 1
    x_values = np.linspace(x_start, x_end, num_steps)
    y_values = y0 * np.exp(coef * x_values)
    return y_values

def errorCalc(y_exact, y_approx):
    local_error = abs(y_exact - y_approx)
    return local_error, sum(local_error)
    
def euler(x_start, x_end, h):
    num_steps = int((x_end - x_start) / h)
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    y_values[0] = 1
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
    return x_values, y_values

x_start, x_end = 0, 2
x1, y_approx1 = euler(x_start, x_end, 0.1)
y_exact1 = exactSolution(x_start, x_end, 0.1, 1, -2)
local_error1, global_error1 = errorCalc(y_exact1, y_approx1)

x2, y_approx2 = euler(x_start, x_end, 0.5)
y_exact2 = exactSolution(x_start, x_end, 0.5, 1, -2)
local_error2, global_error2 = errorCalc(y_exact2, y_approx2)

x3, y_approx3 = euler(x_start, x_end, 1)
y_exact3 = exactSolution(x_start, x_end, 1, 1, -2)
local_error3, global_error3 = errorCalc(y_exact3, y_approx3)

x4, y_approx4 = euler(x_start, x_end, 1.5)
y_exact4 = exactSolution(x_start, x_end, 1.5, 1, -2)
local_error4, global_error4 = errorCalc(y_exact4, y_approx4)

plt.plot(x1, local_error1, label='h = 0.1')
plt.plot(x2, local_error2, label='h = 0.5')
plt.plot(x3, local_error3, label='h = 1')
plt.plot(x4, local_error4, label='h = 1.5')
plt.legend()
plt.xlabel('x')
plt.ylabel('local error at x')
plt.title('local errors')
plt.show()

plt.plot(np.array([0.1, 0.5, 1, 1.5]), np.array([global_error1, global_error2, global_error3, global_error4]))
plt.xlabel('step size')
plt.ylabel('global error')
plt.title('global error as a function of step size')
plt.show()
```
