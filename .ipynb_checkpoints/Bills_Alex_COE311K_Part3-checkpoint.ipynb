```python
import numpy as np
import matplotlib.pyplot as plt

def f(t, x):
    """
    Returns derivatives at a time and position:
    Given: time(t), postition(x)
    """
    return -5000*x + 0.002*np.sin(t)

def g_func(y_next, y_n, t_next, h):
    return y_next - y_n - h * f(t_next, y_next)

def g_prime_func(h):
    return 1 + 5000*h

def backward_euler_step(x, t, h, tol=1e-6, max_iter=20):
    """
    Returns next step and number of iterations:
    Given: postion(x), time(t), step size(h)
    """
    # Initial guess
    x_next = x
    #keeps within max iterations
    for i in range(max_iter):
        g_val = x_next - x - h * f(t+h, x_next)
        g_prime_val = 1 + 5000*h
        # Avoid division by zero
        if abs(g_prime_val) < tol:
            break
        x_new = x_next - g_val / g_prime_val
        # Converged
        if abs(x_new - x_next) < tol:
            return x_new, i+1
        x_next = x_new
    return x_next, max_iter

def backward_euler(t0, tf, h, x0):
    """
    Returns Backward Euler times, postitions, average iterations per step, and the iterations each step took
    Given: initial time (t0), final time(tf), step size(h), initial position(x0)
    """
    #establishes vectors
    n = int((tf - t0)/h)
    t_vals = np.linspace(t0, tf, n+1)
    x_vals = np.zeros(n+1)
    x_vals[0] = x0
    #sets up iteration calcs
    total_iterations = 0
    iterations_per = np.zeros(n+1)
    #calculates each step
    for k in range(n):
        x_vals[k+1], iterations = backward_euler_step(x_vals[k], t_vals[k], h)
        total_iterations += iterations
        iterations_per[k] = iterations
    #averages iterations per step
    avg_iterations = total_iterations / n
    return t_vals, x_vals, avg_iterations, iterations_per

def damped_newton_raphson_step(y_n, t_next, h, g_func, g_prime_func, tol=1e-8, max_iter=20, alpha_min=1e-4):
    """
    Damped Newton-Raphson with line search
    """
    y_guess = y_n
    backtracks_per_iter = []
    alphas_used = []
    damping_used = False
    for k in range(max_iter):
        g = g_func(y_guess, y_n, t_next, h)
        if abs(g) < tol:
            return y_guess, k, backtracks_per_iter, alphas_used, damping_used
        g_prime = g_prime_func(h)
        delta = -g / g_prime
        # Line search for damping factor
        alpha = 1.0
        g_current = abs(g)
        backtracks = 0
        for _ in range(10):  # max 10 backtracking attempts
            y_new = y_guess + alpha * delta
            g_new = abs(g_func(y_new, y_n, t_next, h))
            if g_new < g_current:
                break
            alpha = alpha / 2.0
            backtracks += 1
            damping_used = True
            if alpha < alpha_min:
                print(f"Warning: alpha too small at iter {k}")
                return y_guess, k, backtracks_per_iter, alphas_used, damping_used
        backtracks_per_iter.append(backtracks)
        alphas_used.append(alpha)
        y_guess = y_new
    return y_guess, max_iter, backtracks_per_iter, alphas_used, damping_used

def backward_euler_damped(t0, tf, h, x0):
    n = int((tf - t0)/h)
    t_vals = np.linspace(t0, tf, n+1)
    x_vals = np.zeros(n+1)
    x_vals[0] = x0

    total_iterations = 0
    total_backtracks = 0
    damping_steps = 0

    for k in range(n):
        x_next, iters, backtracks, alphas, used = damped_newton_raphson_step(x_vals[k], t_vals[k] + h, h, g_func, g_prime_func)
        x_vals[k+1] = x_next
        total_iterations += iters
        total_backtracks += sum(backtracks)
        damping_steps += int(used)
    avg_iterations = total_iterations / n
    return t_vals, x_vals, avg_iterations, total_backtracks, damping_steps

h_large = 0.1
t0, tf = 0, 0.3
x0 = 1
# Undamped
t_u, x_u, avg_iterations_u, iterations_per = backward_euler(t0, tf, h_large, x0)
# Damped
t_d, x_d, avg_iterations_d, total_backtracks, damping_steps = backward_euler_damped(t0, tf, h_large, x0)
print("Undamped avg iterations:", avg_iterations_u)
print("Damped avg iterations:", avg_iterations_d)
print("Total backtracks:", total_backtracks)
print("Steps where damping used:", damping_steps)
plt.plot(t_u, x_u, label="Undamped Newton")
plt.plot(t_d, x_d, label="Damped Newton")
plt.xlabel("t")
plt.ylabel("x(t)")
plt.title("Effect of Damping on Newton Convergence (Large h)")
plt.legend()
plt.grid()
plt.show()
```

Damping improves the robustness of Newton-Raphson by scaling the update step using a factor α chosen via line search. Instead of always taking the full step, the algorithm reduces α until a decrease in the residual is observed. This prevents overshooting and divergence, especially in stiff nonlinear problems or when large time steps are used in implicit methods such as Backward Euler.

In comparison, standard Newton-Raphson may require more iterations or fail to converge when the initial guess is poor, whereas the damped version ensures monotonic convergence of the residual at the cost of occasional backtracking.

```python
def step_doubling(x, t, h):
    # one full step
    y_h, it1 = backward_euler_step(x, t, h)
    # first half step
    y_half1, it2 = backward_euler_step(x, t, h/2)
    # second half step
    y_half2, it3 = backward_euler_step(y_half1, t + h/2, h/2)
    err = abs(y_half2 - y_h)
    total_iters = it1 + it2 + it3
    return y_half2, err, total_iters

def adaptive_backward_euler(t0, tf, h0, x0, tol=1e-4, h_min=1e-6, h_max=0.1, safety=0.9):
    t = t0
    x = x0
    h = h0
    t_vals = [t]
    x_vals = [x]
    h_vals = [h]
    accepted = 0
    rejected = 0
    total_iters = 0
    errors = []
    while t < tf:
        if t + h > tf:
            h = tf - t
        x_new, err, iters = step_doubling(x, t, h)
        total_iters += iters
        # accept step
        if err <= tol:
            t += h
            x = x_new
            t_vals.append(t)
            x_vals.append(x)
            h_vals.append(h)
            accepted += 1
            # increase step size safely
            if err == 0:
                h = min(h_max, 2*h)
            else:
                h = min(h_max, safety * h * (tol / err)**0.5)
        # reject step
        else:
            rejected += 1
            h = max(h_min, safety * h * (tol / err)**0.5)
        errors.append(err)
    return np.array(t_vals), np.array(x_vals), np.array(h_vals), accepted, rejected, total_iters, np.array(errors)

t0, tf = 0, 0.1
x0 = 1
h0 = 0.01
t, x, h_used, acc, rej, iters, errors = adaptive_backward_euler(t0, tf, h0, x0)
plt.figure()
plt.plot(t, x)
plt.title("Adaptive Backward Euler Solution")
plt.xlabel("t")
plt.ylabel("x(t)")
plt.grid()
plt.show()
print("Accepted steps:", acc)
print("Rejected steps:", rej)
print("Total Newton iterations:", iters)
plt.figure()
plt.plot(h_used)
plt.title("Adaptive Step Size h over Time")
plt.xlabel("Step index")
plt.ylabel("h")
plt.grid()
plt.show()
```

Step doubling is a technique used to estimate the local truncation error of a numerical method without requiring the exact solution. The idea is to compute the solution over a single time step (h) and compare it with a more accurate solution obtained by taking two successive half steps (h/2). Because the smaller step size produces a more accurate result, the difference between the one step solution and the two step solution provides an estimate of the numerical error. This error estimate is then used to determine whether the step size is appropriate. If the error is too large, the step is rejected and the step size is reduced. If the error is sufficiently small, the step is accepted and the method may increase the step size to improve efficiency.

```python
t0, tf = 0, 0.1
x0 = 1
#adaptive
t_a, x_a, h_used, acc, rej, iters_a, errors_a = adaptive_backward_euler(t0, tf, 0.01, x0)
#fixed step
h_fixed = 0.001
t_f, x_f, iters_f, _ = backward_euler(t0, tf, h_fixed, x0)
x_ref = x_f
t_ref = t_f

def final_error(t, x, t_ref, x_ref):
    x_interp = np.interp(t_ref, t, x)
    return np.max(np.abs(x_interp - x_ref))

err_fixed = final_error(t_f, x_f, t_ref, x_ref)
err_adapt = final_error(t_a, x_a, t_ref, x_ref)
print("\nSAME TARGET ACCURACY")
print("Fixed steps:", len(t_f))
print("Adaptive steps:", len(t_a))
print("Fixed Newton iterations:", iters_f*len(t_f))
print("Adaptive Newton iterations:", iters_a)
print("Fixed final error:", err_fixed)
print("Adaptive final error:", err_adapt)
budget = len(t_a)
h_budget = (tf - t0) / budget
t_b, x_b, iters_b, _ = backward_euler(t0, tf, h_budget, x0)
err_budget_fixed = final_error(t_b, x_b, t_ref, x_ref)
err_budget_adapt = err_adapt


#run adaptive euler
t, x, h_used, acc, rej, iters, errors = adaptive_backward_euler(t0=0, tf=0.1, h0=0.01, x0=1)

#solution with adaptive time points
plt.figure()
plt.plot(t, x, label="Adaptive solution")
plt.scatter(t, x, s=20, color="red", label="Time points")
plt.title("Adaptive Backward Euler Solution")
plt.xlabel("t")
plt.ylabel("x(t)")
plt.legend()
plt.grid()
plt.show()

#step size h(t)
plt.figure()
plt.plot(h_used, marker='o')
plt.title("Adaptive Step Size h Over Time")
plt.xlabel("Step index")
plt.ylabel("h")
plt.grid()
plt.show()


#error estimate vs. time
plt.figure()
plt.plot(errors)
plt.title("Error Estimate vs Step")
plt.xlabel("Step index")
plt.ylabel("Error")
plt.grid()
plt.show()


#accepted vs rejected steps
plt.figure()
plt.bar(["Accepted", "Rejected"], [acc, rej])
plt.title("Accepted vs Rejected Steps")
plt.ylabel("Count")
plt.grid(axis="y")
plt.show()
```

Damping becomes necessary when Newton-Raphson is applied to stiff or highly nonlinear implicit systems where a full Newton step can overshoot and increase the residual instead of reducing it. In these cases, especially with large time steps or poor initial guesses, undamped Newton iterations may oscillate or diverge entirely. Introducing damping with a line search stabilizes convergence by ensuring each update produces a decrease in the residual, making the solver significantly more robust at the cost of a few extra function evaluations.

Adaptive step sizing is worth the added complexity when the solution exhibits multiple time scales, as in stiff ODEs. Fixed step methods either become inefficient (if chosen too small for accuracy everywhere) or unstable (if chosen too large for stiff regions). Adaptive methods automatically refine the step size where the solution changes rapidly and enlarge it where the solution is smooth, achieving a better balance between accuracy and computational cost. This is especially important when solving over long time intervals or when stiffness varies over time.

Overall, the key lesson is that robustness in numerical ODE solvers comes from combining mulitiple ideas. Damping stabilizes nonlinear solves at each step, while adaptive step control stabilizes the integration process across time. Together, they turn Backward Euler from a potentially fragile implicit method into a reliable tool for stiff systems, with predictable accuracy and controlled computational cost.
