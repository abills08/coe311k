```python
import numpy as np
import time
import matplotlib.pyplot as plt

def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C

numRC = 3
timeBefore = 0
naiveRuntime = np.zeros(8)
numpyRuntime = np.zeros(8)
for i in range (0,8):
    X1 = np.random.rand(numRC*(2**i), numRC*(2**i))
    Y1 = np.random.rand(numRC*(2**i), numRC*(2**i))
    timeBefore = time.time()
    naive_matrix_multiplication(X1, Y1)
    naiveRuntime[i] = time.time()-timeBefore
    timeBefore = time.time()
    X1 @ Y1
    numpyRuntime[i] = time.time()-timeBefore
plt.plot(list(range(1,9)), naiveRuntime, color="blue", label="naive")
plt.plot(list(range(1,9)), numpyRuntime, color="red", label="optimized")
plt.legend()
plt.show()
```
